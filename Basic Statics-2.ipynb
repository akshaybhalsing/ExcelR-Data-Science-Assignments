PROBLEM STATEMENT : HOSPITAL PATIENT DATA ANALYSIS

A hospital maintains patient records including admission details, department, diagnosis, doctor, and bill amount. You have two datasets: one with patient info and another with billing details. Some patients have blank bill amounts, and there are multiple rows for the same patient due to follow-ups.

1.Load the patient dataset and show summary with info().

```python
import pandas as pd
patient_df = pd.read_csv("Patient_Data.csv")
patient_df.info()
```

2.Select only the columns relevant for billing:

```python
billing_df = patient_df[['PatientID','Department','Doctor','BillAmount']]
billing_df.head()
```

3.Drop administrative columns like ['ReceptionistID', 'CheckInTime'].

```python
patient_df = patient_df.drop(columns=['ReceptionistID','CheckInTime'])
patient_df.head()
```

4.Use groupby to find total bill amount per department.

```python
department_bill = patient_df.groupby('Department')['BillAmount'].sum()
department_bill
```

5.Remove duplicate patient records based on PatientID.

```python
patient_df = patient_df.drop_duplicates(subset='PatientID')
patient_df.head()
```

6.Fill missing BillAmount values with the mean bill amount.

```python
patient_df['BillAmount'] = patient_df['BillAmount'].fillna(patient_df['BillAmount'].mean())
patient_df.head()
```

7.Merge the billing dataset with patient dataset on PatientID.

```python
billing_data = pd.read_csv("Billing_Data.csv")
merged_df = pd.merge(patient_df, billing_data, on='PatientID')
merged_df.head()
```

8.Concatenate an additional DataFrame that contains new patients for the current week (row-wise).

```python
new_patients = pd.DataFrame({
    'PatientID':[106,107],
    'Name':['Riya','Karan'],
    'Department':['Cardiology','Neurology'],
    'Doctor':['Dr. Mehta','Dr. Singh'],
    'BillAmount':[4500,5200]
})
final_df = pd.concat([merged_df,new_patients],ignore_index=True)
final_df.head()
```

9.Concatenate new billing category columns like ['InsuranceCovered', 'FinalAmount'] (column-wise).

```python
extra_columns = pd.DataFrame({
    'InsuranceCovered':[1000]*len(final_df),
    'FinalAmount':final_df['BillAmount'].fillna(0)-1000
})

final_df = pd.concat([final_df,extra_columns],axis=1)
final_df.head()
```

Expected Outcome:
Final cleaned dataset with accurate billing info.
All missing values handled, merged dataset across PatientID.
Ability to perform further analytics on department-wise revenue or doctor performance.


CONCLUSION:
The patient dataset was successfully cleaned by removing duplicate records, filling missing bill amounts, and dropping unnecessary columns. The patient and billing datasets were merged on PatientID, and additional rows and columns were concatenated successfully. The final dataset is clean and ready for further analysis such as department-wise revenue and doctor performance.
