```python
for col in df.select_dtypes(include=['object']):
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
from sklearn.preprocessing import StandardScaler
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
scaler = StandardScaler()
df_standard = df.copy()
df_standard[num_cols] = scaler.fit_transform(df_standard[num_cols])
df_standard[num_cols].head()
```

```python
from sklearn.preprocessing import MinMaxScaler
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
scaler = MinMaxScaler()
df_minmax = df.copy()
df_minmax[num_cols] = scaler.fit_transform(df_minmax[num_cols])
df_minmax[num_cols].head()
```

--Standard Scaling
Standard Scaling transforms the data so that it has a mean of 0 and a standard deviation of 1.
It is preferred when the data is approximately normally distributed.
It works well with algorithms such as Logistic Regression, Support Vector Machine (SVM), and K-Nearest Neighbors (KNN).
--Min-Max Scaling
Min-Max Scaling rescales the data to a fixed range of 0 to 1.
It is preferred when the features have different ranges and a fixed scale is required.
It is commonly used with Neural Networks and other distance-based algorithms.

●Apply One-Hot Encoding to categorical variables with less than 5 categories.
●Use Label Encoding for categorical variables. Data Exploration and Preprocessing:
●Load the dataset and conduct basic data exploration (summary statistics, missing values, data types).
●les with more than 5 categories.
●Discuss the pros and cons of One-Hot Encoding and Label Encoding.

```python
cat_cols = df.select_dtypes(include=['object']).columns
for col in cat_cols:
    print(col, ":", df[col].nunique())
```

```python
onehot_cols = [col for col in cat_cols if df[col].nunique() < 5]

df_onehot = pd.get_dummies(df, columns=onehot_cols)

df_onehot.head()
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

df_label = df.copy()

label_cols = [col for col in cat_cols if df[col].nunique() >= 5]

for col in label_cols:
    df_label[col] = label_encoder.fit_transform(df_label[col])

df_label.head()
```

--Pros of One-Hot Encoding
Does not create an ordinal relationship between categories.
Suitable for nominal categorical data.
Improves model performance for many machine learning algorithms.
--Cons of One-Hot Encoding
Increases the number of columns.
Can consume more memory for datasets with many categories.
--Pros of Label Encoding
Simple and memory efficient.
Suitable for ordinal categorical data.
Keeps the number of features unchanged.
--Cons of Label Encoding
Introduces an artificial order among categories.
May reduce model performance for nominal data.

3. Feature Engineering:
●Create at least 2 new features that could be beneficial for the model. Explain the rationale behind your choices.
●Apply a transformation (e.g., log transformation) to at least one skewed numerical feature and justify your choice.

```python
df["total_capital"] = df["capital_gain"] - df["capital_loss"]
```

```python
df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 25, 40, 60, 100],
    labels=["Young", "Adult", "Middle_Aged", "Senior"]
)

df[["age", "age_group"]].head()
```

```python
import numpy as np
df["capital_gain_log"] = np.log1p(df["capital_gain"])
df[["capital_gain", "capital_gain_log"]].head()
```

--New Feature 1: Total Capital
Created by subtracting capital_loss from capital_gain.
It represents the overall capital outcome of an individual.
--New Feature 2: Age Group
Created by grouping people into different age categories.
It helps the model identify patterns based on age ranges instead of individual ages.
--Log Transformation
Applied to capital_gain because it is highly skewed.
Log transformation reduces skewness and makes the data more suitable for machine learning models.
