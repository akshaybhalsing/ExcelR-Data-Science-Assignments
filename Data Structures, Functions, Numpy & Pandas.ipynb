1:Create a list of 5 integers. Perform and print the result of the following operations: append, extend, insert, remove, pop, clear, index, count, sort, and reverse.

```python
numbers = [10, 20, 30, 40, 50]
print("Original List:", numbers)

numbers.append(60)
print("After Append:", numbers)

numbers.extend([70, 80])
print("After Extend:", numbers)

numbers.insert(2, 25)
print("After Insert:", numbers)

numbers.remove(40)
print("After Remove:", numbers)

numbers.pop()
print("After Pop:", numbers)

print("Index of 30:", numbers.index(30))

print("Count of 20:", numbers.count(20))

numbers.sort()
print("After Sort:", numbers)

numbers.reverse()
print("After Reverse:", numbers)

numbers.clear()
print("After Clear:", numbers)
```

2:Create a tuple that stores 3 student names. Try changing the second name in the tuple. What happens? Explain why.

```python
students = ("Akshay", "Rahul", "Amit")
print("Original Tuple:", students)
try:
    students[1] = "Rohan"
except TypeError as e:
    print("Error:", e)

print("Explanation: Tuples are immutable, so their elements cannot be changed.")
```

Because tuples are immutable, we cannot modify, add, or remove elements after a tuple is created. Attempting to change an element results in a TypeError.

7:Given a number, check if it's positive, and if it is also even. If not positive, print if it's zero or negative.

```python
number = 6
if number > 0:
    print("The number is positive.")
    if number % 2 == 0:
        print("The number is even.")
    else:
        print("The number is odd.")
elif number == 0:
    print("The number is zero.")
else:
    print("The number is negative.")
```

11:Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'. Then:
- Use .info(), .describe()
- Select 'Name' and 'City' columns
- Drop 'City' column
- Fill any missing values in 'Salary' column with the mean
- Remove any duplicate rows

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Akshay", "Rahul", "Amit", "Neha", "Pooja",
             "Rohan", "Sneha", "Karan", "Priya", "Om"],
    "Age": [21, 22, 20, 23, 21, 24, 22, 23, 20, 21],
    "City": ["Pune", "Mumbai", "Nashik", "Nagpur", "Pune",
             "Delhi", "Mumbai", "Pune", "Nashik", "Pune"],
    "Salary": [30000, 35000, np.nan, 40000, 38000,
               42000, 36000, np.nan, 39000, 30000]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

print("\nDataFrame Info:")
print(df.info())

print("\nDescription:")
print(df.describe())

print("\nName and City Columns:")
print(df[['Name', 'City']])

df = df.drop('City', axis=1)

print("\nAfter Dropping City Column:")
print(df)

df['Salary'] = df['Salary'].fillna(df['Salary'].mean())

print("\nAfter Filling Missing Salary:")
print(df)

df = df.drop_duplicates()

print("\nAfter Removing Duplicate Rows:")
print(df)
```
