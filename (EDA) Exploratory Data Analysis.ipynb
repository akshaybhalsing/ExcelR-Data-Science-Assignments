```python
plt.figure(figsize=(12,6))
sns.boxplot(data=df)
plt.xticks(rotation=90)
plt.show()
```

2.Statistical Summary:

```python
df.describe()
```

```python
print("Mean")
print(df.mean(numeric_only=True))

print("\nMedian")
print(df.median(numeric_only=True))

print("\nStandard Deviation")
print(df.std(numeric_only=True))
```

Interesting Findings:

• The statistical summary provides the mean, median, standard deviation, minimum, maximum, and quartile values for each numerical variable.

• These measures help understand the distribution and variability of the data.

• Variables with high standard deviation indicate greater variation, while variables with similar mean and median values suggest a more symmetric distribution.

3.Data Visualization:

```python
df.hist(figsize=(15,10))
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(data=df)
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.show()
```

```python
sns.pairplot(df.iloc[:,0:5])
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.violinplot(data=df[['LB','AC','FM']])
plt.show()
```
