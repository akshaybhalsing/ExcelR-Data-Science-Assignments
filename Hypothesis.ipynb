BACKGROUND :

Bombay Hospitality Ltd. operates a franchise model for producing exotic Norwegian dinners throughout New England. The operating cost for a franchise in a week (W) is given by the equation:
W = $1,000 + $5X
where X represents the number of units produced in a week. Recent feedback from restaurant owners suggests that this cost model may no longer be accurate, as their observed weekly operating costs are higher.

OBJECTIVE :

To investigate the restaurant owners' claim about the increase in weekly operating costs using hypothesis testing.

DATA PROVIDED

```python


• The theoretical weekly operating cost model:
  W = $1,000 + $5X

• Sample of 25 restaurants with a mean weekly cost of Rs. 3,050

• Number of units produced in a week (X) follows a normal distribution with:
  - Mean (μ) = 600 units
  - Standard Deviation (σ) = 25 units
```

ASSIGNMENT TASK :

1. State the Hypotheses statement:

```python
Null Hypothesis (H₀):
The weekly operating cost is equal to the theoretical cost model.
H₀: μ = 4000
Alternative Hypothesis (H₁):
The weekly operating cost is higher than the theoretical cost model.
H₁: μ > 4000
```

2. Calculate the Test Statistic:

```python
import math
x_bar = 3050      
mu = 1000 + (5 * 600)   
sigma = 5 * 25          
n = 25                  
z = (x_bar - mu) / (sigma / math.sqrt(n))
print("Theoretical Mean (μ):", mu)
print("Population Standard Deviation (σ):", sigma)
print("Test Statistic (Z):", z)
```

3. Determine the Probability and compare:

```python
from scipy.stats import norm
alpha = 0.05
critical_value = norm.ppf(1 - alpha)
p_value = 1 - norm.cdf(z)
print("Critical Value:", critical_value)
print("P-value:", p_value)
```

Interpretation:
• Significance level (α) = 0.05
• Compare the calculated Z-value with the critical value.
• Also compare the P-value with α.

4. Make a Decision:

```python
if z > critical_value:
    print("Decision: Reject the Null Hypothesis (H₀)")
else:
    print("Decision: Fail to Reject the Null Hypothesis (H₀)")
```

Decision:
Compare the calculated Z-test statistic with the critical value.
• If Z > Critical Value, Reject the Null Hypothesis (H₀).
• Otherwise, Fail to Reject the Null Hypothesis (H₀).

5. Conclusion:

Based on the hypothesis test, the decision was made by comparing the calculated Z-test statistic with the critical value and the P-value with the significance level (α = 0.05).
If the Null Hypothesis (H₀) is rejected, there is sufficient evidence to support the restaurant owners' claim that the weekly operating costs are higher than the theoretical cost model.
Otherwise, there is not enough evidence to conclude that the weekly operating costs are higher than the model suggests.
