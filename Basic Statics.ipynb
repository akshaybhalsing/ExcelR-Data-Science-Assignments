```python
import pandas as pd

df = pd.read_csv("sales_data_with_discounts (1).csv")

print(df.head())
```

Descriptive Analytics for Numerical Columns

	Objective: To compute and analyze basic statistical measures for numerical columns in the dataset.
	Steps:
	Load the dataset into a data analysis tool or programming environment (e.g., Python with pandas library).
	Identify numerical columns in the dataset.
	Calculate the mean, median, mode, and standard deviation for these columns.
	Provide a brief interpretation of these statistics.


```python
numerical_columns = df.select_dtypes(include=['number'])

print("Mean")
print(numerical_columns.mean())

print("\nMedian")
print(numerical_columns.median())

print("\nMode")
print(numerical_columns.mode())

print("\nStandard Deviation")
print(numerical_columns.std())
```

Interpretation
Mean represents the average value of each numerical column.
Median represents the middle value and is less affected by outliers.
Mode represents the most frequently occurring value.
Standard deviation measures the spread of the data. A higher value indicates greater variability.

Data Visualization

●	Histograms:
	Plot histograms for each numerical column.
	Analyze the distribution (e.g., skewness, presence of outliers) and provide inferences.


```python
import matplotlib.pyplot as plt
df.hist(figsize=(12,10))
plt.tight_layout()
plt.show()
```

Inference:
• Most numerical columns show different distributions.
• Some columns may be positively skewed.
• A few columns contain extreme values (outliers).

●	Box Plots:
	Create boxplots for numerical variables to identify outliers and the interquartile range.
	Discuss any findings, such as extreme values or unusual distributions.


```python
import matplotlib.pyplot as plt
numerical_columns = df.select_dtypes(include=['number'])
numerical_columns.boxplot(figsize=(12,6))
plt.xticks(rotation=45)
plt.show()
```

Inference:
• Boxplots help identify outliers.
• Some numerical columns contain extreme values.
• Most values lie within the interquartile range.

●	Bar Chart Analysis for Categorical Column:
	Identify categorical columns in the dataset.
	Create bar charts to visualize the frequency or count of each category.
	Analyze the distribution of categories and provide insights.


```python
import matplotlib.pyplot as plt
df['Brand'].value_counts().plot(kind='bar')
plt.title("Brand Distribution")
plt.xlabel("Brand")
plt.ylabel("Count")
plt.show()
```

Inference:
• The chart shows the frequency of each brand.
• Some brands have higher sales records than others.
